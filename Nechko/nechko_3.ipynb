### Виконання базових вправ

Ознайомтеся з базовим прикладом коду в 06-text-generation-apps/python/githubmodels-assignment_ua.ipynb.

Використовуйте ChatCompletionsClient для генерації тексту.

Виконайте тестування з різними параметрами (температура, кількість токенів).


```python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential
from dotenv import load_dotenv

load_dotenv()
token = os.getenv('GITHUB_TOKEN_MODEL')

endpoint = "https://models.inference.ai.azure.com"

model_name = "gpt-4o"

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

prompt = "Покажи мені 5 рецептів для страви з наступними інгредієнтами: курка, картопля та морква. Для кожного рецепту перелічи всі використані інгредієнти"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "Ти корисний асистент.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Було:
    # temperature=1.,
    # max_tokens=1000,
    # Стало
    temperature=1,
    max_tokens=300,
    top_p=1.
)

print(response.choices[0].message.content)
```

### Індивідуальне завдання (Варіант №3)
Розробіть історичного бота, що відтворює манеру мовлення історичних постатей.

```python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential
from dotenv import load_dotenv

load_dotenv()
token = os.getenv('GITHUB_TOKEN_MODEL')

endpoint = "https://models.inference.ai.azure.com"

model_name = "gpt-4o"

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

# Історичний бот - Богдан Хмельницький
user_question = input("Напиши додаткові теми для включення у відповідь:")

prompt = f"Розкажи про важливість козацької звитяги для української державності. Також включи інформацію про: {user_question}"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "Ти - гетьман Війська Запорозького Богдан Хмельницький. Говори стилізованою староукраїнською мовою XVII століття, використовуй історичні терміни та характерні для того часу вирази. Тримайся гордовитої, але мудрої манери поведінки козацького ватажка.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    temperature=0.8,
    max_tokens=800,
    top_p=0.9
)

print(response.choices[0].message.content)
```
